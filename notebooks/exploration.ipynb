```python
import pandas as pd
from electricity_price_forecasting.data.data_pipeline import fetch_data

pd.set_option("display.max_columns", None)
pd.set_option("display.expand_frame_repr", False)
pd.set_option("display.max_colwidth", None)
```

```python
merged_df, quarterly_df, hourly_df = fetch_data(pd.Timestamp("2024-07-01"), pd.Timestamp("2025-03-06"))
```

```python
display(merged_df.head(), quarterly_df.head(), hourly_df.head())
```

```python
display(quarterly_df.describe(), hourly_df.describe())
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import mean_squared_error

for date in pd.date_range("2024-07-01", "2025-03-06", freq="2W", inclusive="left", tz="Europe/Prague"):
    df = hourly_df[date:date + pd.DateOffset(weeks=2)].dropna()

    id_price = df["ID Weighted average of the prices (EUR/MWh)"]
    da_price = df["DA Price (EUR/MWh)"]

    plt.figure(figsize=(12, 6))
    plt.plot(df.index, id_price, marker="o", label="Intra-day average price", alpha=0.6, markersize=1)
    plt.plot(df.index, da_price, linestyle="--", label="Day-ahead predicted price")
    plt.xlabel("Time")
    plt.ylabel("Price (EUR/MWh)")
    plt.title("Comparison of Intra-day average price and Day-ahead predicted price")
    plt.legend()
    plt.grid()

    rmse = np.sqrt(mean_squared_error(da_price, id_price))
    print(f"Root Mean Square Error (RMSE): {rmse:.4f}")

    plt.show()
```
