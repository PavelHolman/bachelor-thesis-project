```python
import pandas as pd
from pandas.errors import PerformanceWarning
from sklearn.exceptions import UndefinedMetricWarning
import warnings
import sys
import os

pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)

warnings.simplefilter('ignore', category=PerformanceWarning)
warnings.simplefilter('ignore', category=UndefinedMetricWarning)
sys.path.append(os.path.abspath(os.path.join(os.getcwd(), "..")))
```

```python
from src.models.train_intra_day_time_series import train_intra_day_time_series
from src.testing.data_split import ID_TARGET
from src.testing.datasets import load_intra_day_time_series_split
from src.testing.evaluation import predict_time_series_model, evaluate
from src.testing.metrics import get_eval_metrics

eval_metrics = get_eval_metrics()
train, val, select, test = load_intra_day_time_series_split()
model = train_intra_day_time_series(train, val, "MAE", cache="MAE")

pred_train = predict_time_series_model(model, train, cache="TS ID MAE Train")
pred_val = predict_time_series_model(model, val, cache="TS ID MAE Val")
pred_select = predict_time_series_model(model, select, cache="TS ID MAE Select")

y_train = pd.Series(train[ID_TARGET].values, index=train.index.get_level_values(1))
y_val = pd.Series(val[ID_TARGET].values, index=val.index.get_level_values(1))
y_select = pd.Series(select[ID_TARGET].values, index=select.index.get_level_values(1))

metrics_train = evaluate(pred_train, y_train, eval_metrics)
metrics_val = evaluate(pred_val, y_val, eval_metrics)
metrics_select = evaluate(pred_select, y_select, eval_metrics)

display(metrics_train, metrics_val, metrics_select)
```
